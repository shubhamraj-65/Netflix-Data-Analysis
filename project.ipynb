## Task 3 — Country-Wise Netflix Content Analysis

- The Top 10 countries show that Netflix content is distributed across multiple international markets.



## Task 4 — Trend Analysis by Release Year

```python
yearly_content = data['release_year'].value_counts().sort_index()

print(yearly_content)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(yearly_content.index, yearly_content.values)

plt.title('Netflix Content Releases by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')

plt.grid(True)
plt.show()
```

```python
yearly_content.idxmax()
```

```python
yearly_content.max()
```

### Trend Analysis

- Netflix content releases show an overall increasing trend over the years.
- Content growth becomes more noticeable after 2000 and rises sharply after 2015.
- The highest number of titles was recorded in 2018, with 1,146 titles.
- After 2018, the number of recorded titles declines in the dataset.

### Key Business Pattern

The dataset shows a strong expansion in Netflix content during the 2010s, with the highest recorded release-year count occurring in 2018.

## Task 5 — Content Rating & Genre Analysis

```python
rating_count = data['rating'].value_counts()

print(rating_count)
```

```python
data['listed_in'].head()
```

```python
data['listed_in'].value_counts()
```

```python
genre_count = (
    data['listed_in']
    .str.split(', ')
    .explode()
    .value_counts()
)

print(genre_count)
```

```python
top_10_genres = genre_count.head(10)

plt.figure(figsize=(10,6))
plt.barh(top_10_genres.index, top_10_genres.values)

plt.title('Top 10 Netflix Genres')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')

plt.gca().invert_yaxis()
plt.show()
```

```python
rating_by_type = pd.crosstab(data['rating'], data['type'])

print(rating_by_type)
```

```python
import plotly.express as px
```

```python
rating_by_type =pd.crosstab(data['rating'],data['type'])
```

```python
fig = px.bar(
    rating_by_type,
    barmode='group',
    title='Netflix Ratings by Content Type',
    labels={
        'value': 'Number of Titles',
        'rating': 'Rating',
        'variable': 'Content Type'
    }
)

fig.show()
```

### Audience Insights

- TV-MA is the most common rating in the dataset, with 3,205 titles overall.
- TV-14 is the second most common rating, with 2,157 titles.
- Movies have a much higher number of TV-MA-rated titles (2,062) compared with TV Shows (1,143).
- TV Shows have higher counts than Movies for TV-Y (175 vs 131) and TV-Y7 (194 vs 139), indicating a stronger presence of youth-oriented ratings among TV Shows.
- International Movies, Dramas, and Comedies are the three most common genre categories in the dataset.
- The rating distribution shows that Netflix content in this dataset is concentrated mainly in mature and teen-oriented rating categories.

## Netflix Business Insights

### 1. Content Type Distribution
- The dataset contains 8,790 Netflix titles.
- Movies account for 6,126 titles (69.69%), while TV Shows account for 2,664 titles (30.31%).
- Movies represent more than twice the number of TV Shows in the dataset.

### 2. Country-Wise Content
- The United States has the highest number of titles with 3,240 records.
- India is the second-highest contributor with 1,057 titles.
- The United Kingdom ranks third with 638 titles.
- The dataset shows content production across multiple countries, indicating a broad international content presence.

### 3. Genre Distribution
- International Movies are the most common genre category with 2,752 titles.
- Dramas are the second most common with 2,426 titles.
- Comedies rank third with 1,674 titles.
- International TV Shows also have a significant presence with 1,349 titles.

### 4. Release-Year Trend
- The number of titles generally increased over the years, particularly after 2000.
- A strong increase is visible after 2015.
- 2018 recorded the highest number of titles in the dataset, with 1,146 releases.
- The number of recorded titles declines after 2018; this pattern reflects the available records in the dataset and should not by itself be interpreted as an actual decline in Netflix production.

### 5. Rating Analysis
- TV-MA is the most common rating with 3,205 titles.
- TV-14 is the second most common rating with 2,157 titles.
- Movies have 2,062 TV-MA titles, compared with 1,143 TV-MA TV Shows.
- TV Shows have higher counts than Movies for TV-Y and TV-Y7 ratings.

### 6. Key Business Takeaways
- The Netflix catalog in this dataset is heavily concentrated in Movies.
- International content categories form a major part of the catalog.
- Drama and Comedy are among the major genre categories.
- Mature and teen-oriented ratings account for a large share of the catalog.
- The strongest concentration of release records occurs during the late 2010s, with 2018 being the peak year in this dataset.

## Executive Summary

This analysis explored 8,790 Netflix titles to understand the platform's content distribution, geographic presence, genre patterns, release trends, and rating categories.

The analysis found that Movies dominate the dataset, representing 69.69% of all titles, while TV Shows account for 30.31%. The United States is the largest content-producing country in the dataset, followed by India and the United Kingdom.

International Movies, Dramas, and Comedies are the most common genre categories. The release-year analysis shows strong growth in recorded content during the 2010s, with 2018 having the highest number of titles at 1,146.

## Conclusion

The analysis provides a clear overview of Netflix's catalog composition and major content patterns. The findings can support content portfolio analysis, international content planning, genre monitoring, audience segmentation, and future trend analysis.

However, this dataset contains catalog-level information only. Therefore, conclusions about viewer demand, revenue, profitability, or audience preferences cannot be directly determined without additional data such as viewing hours, ratings, revenue, or user engagement.
